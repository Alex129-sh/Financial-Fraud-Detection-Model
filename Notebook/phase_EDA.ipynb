```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
# LOAD CLEANED DATA
df=pd.read_csv("fraud_cleaned.csv")

print("Shape:",df.shape)
print("\nTarget Distribution:")
print(df['Fraudulent'].value_counts())
print(df['Fraudulent'].value_counts(normalize=True).round(4)*100)
```

```python
# SET STYLE
sns.set(style="whitegrid")
plt.rcParams['figure.figsize']=(10,6)
```

```python
#1 TARGET DISTRIBUTION
plt.figure(figsize=(8,4))
sns.countplot(x='Fraudulent', data=df, palette='Set2')
plt.title("Fraud vs Non-Fraud Distribution")
plt.show()
```

```python
#2.TRANSACTION AMOUNT DISTRIBUTION
plt.figure(figsize=(10,4))
sns.histplot(data=df, x='Transaction_Amount', hue='Fraudulent', bins=50, kde=True)
plt.title("Transaction Amount Distribution by Fraud")
plt.show()
```

```python
#3 BOXPLOT OF AMOUNT VS FRAUD
plt.figure(figsize=(8,5))
sns.boxplot(x='Fraudulent', y='Transaction_Amount', data=df)
plt.title("Transaction Amount vs Fraud")
plt.show()
```

```python
# 4. Fraud by Merchant Category
plt.figure(figsize=(12,5))
sns.countplot(data=df, x='Merchant_Category', hue='Fraudulent')
plt.xticks(rotation=45)
plt.title("Fraud by Merchant Category")
plt.show()
```

```python
# 5. Fraud by Device Type
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Device_Type', hue='Fraudulent')
plt.title("Fraud by Device Type")
plt.show()
```

```python
# 6. Fraud by Payment Method
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='Payment_Method', hue='Fraudulent')
plt.xticks(rotation=45)
plt.title("Fraud by Payment Method")
plt.show()
```

```python
# 7. Fraud by Location
plt.figure(figsize=(12,5))
sns.countplot(data=df, x='Location', hue='Fraudulent')
plt.xticks(rotation=45)
plt.title("Fraud by Location")
plt.show()
```

```python
# 7. Fraud by Location
plt.figure(figsize=(12,5))
sns.countplot(data=df, x='Location', hue='Fraudulent')
plt.xticks(rotation=45)
plt.title("Fraud by Location")
plt.show()
```

```python
# 8. International Transactions vs Fraud
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Is_International', hue='Fraudulent')
plt.title("International Transactions vs Fraud")
plt.show()
```

```python
# 9. Suspicious Keyword vs Fraud
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Suspicious_Keyword', hue='Fraudulent')
plt.title("Suspicious Keyword vs Fraud")
plt.show()
```

```python
# 10. Correlation Heatmap (Numerical columns)
plt.figure(figsize=(10,7))
num_cols = ['Transaction_Amount', 'Previous_Transactions', 'Average_Spend', 
            'Account_Age_Days', 'Is_International', 'Fraudulent']
sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()

print("\n EDA Completed!")
```
