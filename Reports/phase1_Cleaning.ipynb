```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
# 1. Load the dataset
df = pd.read_csv("financial_fraud_detection_dataset.csv")
print("Original Shape:", df.shape)
print("\nColumn Names:\n", df.columns.tolist())
print("\nData Types:\n", df.dtypes)
print("\nMissing Values:\n", df.isnull().sum())
print("\nTarget Distribution:\n", df['Fraudulent'].value_counts(normalize=True))
```

```python
# 2. Convert Transaction_Date to datetime
df['Transaction_Date'] = pd.to_datetime(df['Transaction_Date'], dayfirst=True)
```

```python
# 3. Check for duplicates
print("\nDuplicate Rows:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
# 4. Basic sanity checks
print("\nNegative Amounts:", (df['Transaction_Amount'] < 0).sum())
print("Zero or Negative Account Age:", (df['Account_Age_Days'] <= 0).sum())
```

```python
# 5. Drop ID columns (not useful for modeling)
df_clean = df.drop(columns=['Transaction_ID', 'Customer_ID'])
```

```python
# 6. Final check
print("\nCleaned Shape:", df_clean.shape)
print("\nCleaned Columns:\n", df_clean.columns.tolist())
print("\nData Types after cleaning:\n", df_clean.dtypes)
```

```python
# 7. Save cleaned dataset
df_clean.to_csv("fraud_cleaned.csv", index=False)
print("\n✅ Cleaned dataset saved as 'fraud_cleaned.csv'")
```
